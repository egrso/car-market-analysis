```python
import pandas as pd
```

```python
df = pd.read_parquet("../data/processed/clean.parquet")
```

```python
df.info()
```

```python
df.isna().sum()
```

Замена строк

```python
df['color'] = df['color'].fillna('unknown')
df['description'] = df['description'].fillna('')
df['is_dealer'] = df['is_dealer'].astype('boolean')
```

Замена числовых значений

```python
df = df.dropna(subset=['fuelType'])
```

```python
df[df.mileage.isna()].year.value_counts().head()
```

```python
df['mileage_missing'] = df['mileage'].isna()
```

```python
df.nunique()
```

```python
df.loc[df['power'].isna(), 'fuelType'].eq('Электро').sum()
```

- `mileage` — пробег отсутствует в 50,5% объявлений, особенно часто у машин 2022–2023 годов. Оснований считать его нулевым нет, а подстановка медианы для половины объявлений исказит распределение. Отсутствие пробега отмечено флагом mileage_missing.
- `mileage_per_year` — среднегодовой пробег рассчитывается из пробега и возраста автомобиля. При неизвестном пробеге или нулевом возрасте рассчитать показатель нельзя.
- `power` — 3 054 из 3 065 пропусков приходятся на электромобили. У некоторых моделей мощность отсутствует во всех объявлениях, а медиана по марке может заметно отличаться от мощности конкретной модели.
